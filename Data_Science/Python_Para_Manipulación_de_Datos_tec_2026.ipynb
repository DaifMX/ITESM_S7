# Tema 7: Bucles para iteración sobre registros
## Teoría Básica
Aunque en Pandas usaremos operaciones vectorizadas (más rápidas), los bucles for son esenciales para:

Leer archivos uno por uno de una carpeta.

Hacer web scraping.

Aplicar lógica compleja que no es vectorizable.

range(): Genera secuencias numéricas.

```python
# Lista de precios crudos recibidos de una API
precios_crudos = [100, 250, None, 400, 150]
precios_con_iva = []

print("Iniciando procesamiento de lista...")

for precio in precios_crudos:
    # Validación simple dentro del bucle
    if precio is not None:
        nuevo_precio = precio * 1.16  # Agregar IVA
        precios_con_iva.append(nuevo_precio)
    else:
        # Estrategia de imputación: llenar con 0 o ignorar
        precios_con_iva.append(0.0)

print(f"Lista original: {precios_crudos}")
print(f"Lista procesada: {precios_con_iva}")
```

# Tema 8: Control avanzado de ciclos (break/continue)
## Teoría Básica
Optimizan el rendimiento de los bucles en procesos ETL (Extract, Transform, Load).

break: Detiene el bucle por completo. Útil cuando buscas un solo dato ("encontrar el primer error") y quieres dejar de procesar para ahorrar recursos.

continue: Salta la iteración actual y pasa a la siguiente. Útil para omitir "datos basura" sin detener todo el proceso.

```python
log_servidor = ["OK", "OK", "WARNING", "OK", "ERROR", "OK", "CRITICAL"]

print("--- Buscando primera falla crítica ---")

for estado in log_servidor:
    if estado == "OK":
        continue  # Salta al siguiente, no nos interesa imprimir los OK

    print(f"Revisando evento: {estado}")

    if estado == "ERROR":
        print("¡Falla encontrada! Deteniendo análisis para enviar alerta.")
        break # Rompe el ciclo, no lee "CRITICAL"

print("Fin del análisis.")
```

# Tema 9: Manejo básico de excepciones
## Teoría Básica
En DataOps, los datos siempre vienen sucios. Si un script se detiene a la mitad de la noche por un error de formato, es un problema. try-except permite manejar errores sin "crashear" el programa.

try: Código propenso a fallar.

except: Qué hacer si falla.

```python
datos_mixtos = ["10", "20", "treinta", "40", "50"]
datos_limpios = []

for item in datos_mixtos:
    try:
        # Intentamos convertir a entero
        numero = int(item)
        datos_limpios.append(numero)
    except ValueError:
        # Si falla la conversión, entramos aquí
        print(f"Advertencia: No se pudo convertir '{item}'. Registro omitido.")
    except Exception as e:
        # Catch-all para errores imprevistos (usar con cuidado)
        print(f"Error desconocido: {e}")

print(f"Datos recuperados: {datos_limpios}")
```

# Tema 10: Estructura y uso de funciones
## Teoría Básica
Una función es un bloque de código con nombre que realiza una tarea específica.

def: Palabra clave para definirla.

Parámetros: Variables que la función recibe para trabajar (inputs).

return: El resultado que la función devuelve (output). Crucial: A diferencia de print (que solo muestra texto), return entrega el dato para que el programa lo siga usando.

Scope (Alcance): Las variables creadas dentro de una función mueren cuando la función termina.

```python
# Definición de la función
def calcular_impuesto(monto, tasa=0.16):
    """
    Calcula el impuesto de un monto dado.
    Si no se especifica tasa, usa 16% por defecto.
    """
    impuesto = monto * tasa
    return impuesto

# Uso de la función
venta_1 = 1000
impuesto_1 = calcular_impuesto(venta_1) # Usa 0.16 por defecto

venta_2 = 2000
impuesto_2 = calcular_impuesto(venta_2, tasa=0.21) # Sobreescribe la tasa

print(f"Impuesto venta 1: ${impuesto_1}")
print(f"Impuesto venta 2: ${impuesto_2}")

# Demostración de return vs print:
total_neto = venta_1 + impuesto_1 # Podemos operar con el resultado porque hubo return
print(f"Total a pagar: ${total_neto}")
```

# Tema 11: Funciones reutilizables para DataOps
## Teoría Básica
En pipelines de datos, usamos funciones para "limpiar" o "transformar" datos crudos repetitivamente.

Type Hinting: def funcion(texto: str) -> str: Ayuda a saber qué tipo de dato entra y sale (vital en equipos de datos).

Manejo de errores interno: La función debe saber qué hacer si recibe basura (ej. retornar None o un valor por defecto).

```python
def limpiar_precio(precio_sucio: str) -> float:
    """
    Recibe un string con símbolos (ej. '$1,200.50'),
    quita caracteres extraños y devuelve un float.
    Retorna -1.0 si hay error.
    """
    if not isinstance(precio_sucio, str):
        return -1.0 # O manejar el error según lógica de negocio

    # Limpieza: quitar $ y comas
    limpio = precio_sucio.replace("$", "").replace(",", "").strip()

    try:
        return float(limpio)
    except ValueError:
        return -1.0

# Simulamos procesar una columna de datos sucia
lista_precios = ["$1,500.00", "  200  ", "Error", "$50.5"]
lista_procesada = []

for item in lista_precios:
    lista_procesada.append(limpiar_precio(item))

print(f"Datos limpios: {lista_procesada}")
```

# Tema 15: Operaciones vectorizadas y estadísticas
## Teoría Básica
Vectorización es la capacidad de aplicar una operación a todo el array sin escribir un bucle for.

Aritmética: array + 5 suma 5 a cada elemento. array1 * array2 multiplica elemento a elemento.

Broadcasting: NumPy "estira" automáticamente arrays de distintos tamaños para poder operarlos (ej. sumar un escalar a una matriz).

Agregación: Métodos optimizados como .sum(), .mean() (promedio), .max(), .std() (desviación estándar).

```python
import numpy as np

# Datos simulados: Precios de 5 productos
precios = np.array([100, 200, 150, 300, 50])

# OPERACIÓN VECTORIZADA (Sin bucle for)
# Aplicar 20% de descuento a TODOS los productos
precios_descuento = precios * 0.80

print(f"Precios originales: {precios}")
print(f"Con descuento:      {precios_descuento}")

# ESTADÍSTICAS BÁSICAS
total_ventas_esperado = precios_descuento.sum()
precio_promedio = precios.mean()
precio_maximo = precios.max()

print(f"Total esperado: ${total_ventas_esperado}")
print(f"Precio promedio (orig): ${precio_promedio}")
```

# Tema 16: Filtrado (Masking) y transformaciones
## Teoría Básica
¿Cómo seleccionamos datos que cumplen una condición? En NumPy usamos Máscaras Booleanas.

Condición: array > 50 devuelve un array de True/False.

Filtrado: array[condicion] devuelve solo los elementos donde la condición es True.

Where: np.where(condicion, valor_si_si, valor_si_no) es como el "SI" (IF) de Excel vectorizado.

```python
import numpy as np

temperaturas = np.array([22, 25, 30, 18, 40, 12, 35])

# 1. Crear una máscara (pregunta lógica)
# ¿Qué días hizo mucho calor (> 28 grados)?
es_caluroso = temperaturas > 28
print(f"Máscara booleana: {es_caluroso}")

# 2. Filtrar usando la máscara
dias_calurosos = temperaturas[es_caluroso]
print(f"Temperaturas altas filtradas: {dias_calurosos}")

# 3. Transformación con np.where
# Etiquetar numéricamente: 1 si es calor (>28), 0 si no
etiquetas = np.where(temperaturas > 28, 1, 0)
print(f"Etiquetas (Dummy variables): {etiquetas}")
```

Problema 31: Filtrar Outliers (Valores atípicos) Tienes un sensor que a veces falla y da valores negativos: lecturas = np.array([10, -5, 12, -1, 15]). Crea un nuevo array lecturas_validas que contenga solo los valores positivos.

```python
lecturas = np.array([10, -5, 12, -1, 15])

# Filtramos donde la lectura sea mayor o igual a 0
lecturas_validas = lecturas[lecturas >= 0]

print(lecturas_validas) # [10, 12, 15]
```

```python
# Solución problema 32
lecturas = np.array([10, -5, 12, -1, 15])
lecturas = np.where(lecturas < 0, 0, lecturas)

print(lecturas)
```

# Tema 19: Limpieza y preparación de datos
## Teoría Básica
El 80% del tiempo se va aquí. Datos "sucios" = Análisis erróneo (GIGO: Garbage In, Garbage Out).

Nulos (NaN): df.isna(), df.dropna() (eliminar fila), df.fillna(valor) (rellenar).

Duplicados: df.duplicated(), df.drop_duplicates().

Tipos incorrectos: df["col"].astype(int).

```python
import pandas as pd
import numpy as np

# DataFrame sucio
df_sucio = pd.DataFrame({
    "Ciudad": ["CDMX", "CDMX", "Monterrey", None, "Guadalajara"],
    "Temperatura": [25, 25, np.nan, 30, 28], # np.nan es un valor nulo
    "ID_Sensor": ["1", "1", "2", "3", "4"]   # Duplicado en fila 1 y 2
})

print("--- Datos Sucios ---")
print(df_sucio)

# 1. Eliminar duplicados
df_limpio = df_sucio.drop_duplicates()

# 2. Rellenar nulos (Imputación)
# Rellenamos temperatura faltante con el promedio
promedio = df_limpio["Temperatura"].mean()
df_limpio["Temperatura"] = df_limpio["Temperatura"].fillna(promedio)

# 3. Eliminar filas donde la Ciudad sea nula (imposible de geolocalizar)
df_limpio = df_limpio.dropna(subset=["Ciudad"])

print("\n--- Datos Limpios ---")
print(df_limpio)
```

# Tema 20: Transformaciones comunes (Filtros, GroupBy, Merges)
## Teoría Básica
Aquí respondemos preguntas de negocio.

Filtros: df[df["col"] > 50].

Agregación (GroupBy): "Tabla dinámica" en código. df.groupby("Categoria")["Ventas"].sum().

Joins (Merge): Unir dos tablas por una clave común (como VLOOKUP/BUSCARV o SQL JOIN). pd.merge(tabla1, tabla2, on="id").

```python
import pandas as pd

# Tabla de Ventas
ventas = pd.DataFrame({
    "Vendedor": ["Ana", "Luis", "Ana", "Luis", "Sol"],
    "Monto": [100, 200, 150, 50, 300],
    "ID_Zona": [1, 1, 2, 2, 1]
})

# Tabla de Zonas (Catálogo)
zonas = pd.DataFrame({
    "ID_Zona": [1, 2],
    "Nombre_Zona": ["Norte", "Sur"]
})

# 1. MERGE (Join): Unir ventas con el nombre de la zona
df_completo = pd.merge(ventas, zonas, on="ID_Zona", how="left")

# 2. GROUP BY: ¿Cuánto vendió cada vendedor?
resumen_vendedores = df_completo.groupby("Vendedor")["Monto"].sum()

# 3. FILTRO: Solo ventas de la Zona Norte
ventas_norte = df_completo[df_completo["Nombre_Zona"] == "Norte"]

print("--- Tabla Unida (Merge) ---")
print(df_completo)
print("\n--- Ventas por Vendedor ---")
print(resumen_vendedores)
```
