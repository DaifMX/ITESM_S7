## Handling missing data

The **isna** function identifies records that contain missing data. This function returns a new DataFrame with only boolean values.

```python
df.isna()
```

The **notna** function identifies the elements that are not missing.

```python
df.notna()
```

You can work with a single column to make decisions for each attribute.

```python
attr = df.Literacy
attr.notna()
```

To remove records with NaN values, you can use **dropna**. This method works with both Series and DataFrames.

```python
df.Literacy.dropna()
```

```python
df.dropna()
```

If, instead of removing missing values, you want to fill them with data, you can use the **fillna** function.

```python
attr.fillna(5000)
```

To perform mean imputation, you need to calculate the mean of the Series without NaN values and then assign it with fillna.

```python
# Calculate the mean
ave = attr.mean()
print("Mean:", ave)

# Assign the mean with fillna
attr.fillna(ave)
```

#### **The functions fillna, dropna, isna, notna, and the [] operator create new DataFrames and new Series. In other words, the original DataFrame or Series is not modified.**

To fill a specific column with fillna, you only need to indicate the name of the column to modify.

```python
# Calculate the means of the attributes to impute
ave = df.Literacy.mean()
print("Mean:", ave)

# Modify only the desired column
df.fillna(value = {'Literacy' : ave})
```

You can also perform imputation for multiple columns at the same time.

```python
# Calculate the means of the attributes to impute
literacy_mean = df.Literacy.mean()
print("Literacy:", literacy_mean)

income_mean = df.Income.mean()
print("Income:", income_mean)

# Modify only the desired columns
df.fillna(value = {'Literacy' : literacy_mean, 'Income' : income_mean})
```

## Outlier detection
To detect possible outliers, it is possible to follow two approaches:
<ul>
    <li>Use a graphical tool to identify outliers and see the values around which they occur. The best-known tool for this is the box-and-whisker plot.</li>
    <li>Use a statistical or analytical method to detect extreme values. The best-known methods for this are the z-score method and the Boxplot rule.</li>
</ul>

The following lines show an example of using the z-score method to detect outliers.

```python
# Z-score method for outlier detection

# Select the data to analyze
data = df.Income

# Remove missing data
data = data.dropna()

# Calculate the mean value and standard deviation
m = data.mean()
s = data.std()

# Determine the limits for outliers
lower_limit = m - 3*s
upper_limit = m + 3*s

# Determine which elements are valid
valid = (data >=lower_limit) & (data <=upper_limit)

# Select the elements that are valid
sel_data = data[valid]
sel_data
```

```python
# Display the out-of-range values
not_valid = (data < lower_limit) | (data > upper_limit)

not_sel_data = data[not_valid]
not_sel_data
```

If the data deviate substantially from a **normal distribution**, it is preferable to use a quartile-based method such as the Box Plot rule.

```python
# Boxplot rule method for detecting outliers

# Select the data to analyze
data = df.Income

# Remove missing data
data = data.dropna()

# Calculate the first and third quartiles of the data
q1 = data.quantile(0.25)
q3 = data.quantile(0.75)

# Calculate the interquartile range
iqr = q3 - q1

# Determine the limits for outliers
lower_limit = q1 - 1.5*iqr
upper_limit = q3 + 1.5*iqr

# Determine which elements are valid
valid = (data >=lower_limit) & (data <=upper_limit)

# Select the elements that are valid
sel_data = data[valid]
sel_data
```

```python
# Display the out-of-range values
not_valid = (data < lower_limit) | (data > upper_limit)

not_sel_data = data[not_valid]
not_sel_data
```

## Descriptive statistics by class (continent)
Because **Continent** is a categorical class, it is possible to describe the data for each class of this attribute.

To find out which values **Continent** takes, use the **unique** function.

```python
cont_values = df.Continent.unique()
cont_values
```

To analyze each continent, simply separate the desired data and then apply the standard DataFrame functions.

```python
# Separate the data for continent 1
df_c1 = df[df.Continent == 1]
df_c1
```

```python
# Display the description of the desired columns
selected_columns = ['Fertility', 'Literacy', 'Income', 'Primary_fem', 'Primary_male', 'Population']
df_c1_red = df_c1[selected_columns]

df_c1_red.describe()
```

The same can be done for all continents. This can be achieved with a **for** loop, or one by one.



```python
for c in cont_values:
    print ('Continent:', c)
    df_c = df[df.Continent == c]
    df_c_red = df_c[selected_columns]
    print(df_c_red.describe())
```

To make categories or classes easier to manage, Pandas provides the **groupby** function, which returns a new DataFrame partitioned into subtables, each representing the data for one class.

```python
selected_columns = ['Continent', 'Fertility', 'Literacy', 'Income', 'Primary_fem', 'Primary_male', 'Population']
df_sel = df[selected_columns]

grouped_df = df_sel.groupby(['Continent'])
grouped_df
```

Grouped DataFrames cannot be displayed easily, but they allow the same operation to be applied to all groups at the same time.

```python
# Display the mean of each DataFrame
grouped_df.mean()
```

```python
# Display the description of each DataFrame
grouped_df.describe()
```

Another feature of grouped data is that you can iterate through each group and work with each group's DataFrame separately.

```python
for name, group in grouped_df:
    print(name)
    print(group)
    print(group.mean())
    print()
```

### Line plot
This is the simplest plot and is used to relate two variables. It is generated by calling the **plot** function.

```python
# Extract the data to plot and remove NaN values
data = df[['Income', 'Literacy']]
data = data.dropna()

# Since this is an X-Y plot, where we want to see the relationship between Income and Literacy, it is necessary to sort the X variable (in this case Income)
data = data.sort_values(by = ['Income'])

# Build the plot. The first argument is X, the second argument is Y.
plt.plot(data.Income, data.Literacy)

# Plot formatting (axes, title, etc.)
plt.xlabel("Income")
plt.ylabel("Literacy")
plt.title("Income Vs Literacy")

# Display the plot
plt.show()
```

More than two plots can be added to the same area by calling the **plot** function more than once.

```python
# Extract the data to plot and remove NaN values
data = df[['Income', 'Primary_fem', 'Primary_male']]
data = data.dropna()

# Since this is an X-Y plot, where we want to see the relationship between Income and Literacy, it is necessary to sort the X variable (in this case Income)
data = data.sort_values(by = ['Income'])

# Build the plot. The first argument is X, the second argument is Y.
plt.plot(data.Income, data.Primary_fem, label = 'Female', color = 'blue')
plt.plot(data.Income, data.Primary_male, label = 'Male', color = 'red')

# Plot formatting (axes, title, etc.)
plt.xlabel("Income")
plt.ylabel("Net primary school enrolment ratio")
plt.title("Income vs Primary school enrolment")

# Add legend()
plt.legend()

# Display the plot
plt.show()
```

## Visualization with Pandas
Pandas has some functions for plotting DataFrames directly. These functions use MatplotLib internally.

Pandas' plotting function is **plot**, which can receive the following arguments.

<ul>
    <li>kind: A string parameter for the type of graph, such as line, bar, barh, hist, box, KDE, pie, area, or scatter.</li>
    <li>figsize: This defines the size for a figure in a tuple of (width, height).</li>
    <li>title: This defines the title for the graph.</li>
    <li>grid: Boolean parameter for the axis grid line.</li>
    <li>legend: This defines the legend.</li>
    <li>xticks: This defines the sequence of x-axis ticks.</li>
    <li>yticks: This defines the sequence of y-axis ticks.</li>
</ul>

In addition, each type accepts certain arguments such as the names of the variables to plot, etc.

One advantage of this type of plot is that subplots can be automatically generated for each attribute or column.

```python
# Scatter plot
df.plot(kind = 'scatter', x = 'Literacy', y = 'Income', color = 'red', title = 'Literacy vs Income')
plt.show()
```
