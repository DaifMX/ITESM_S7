# Comparación de resultados

Ejecuta primero los notebooks de PyTorch, scikit-learn y Keras. Este notebook reúne sus métricas, curvas de pérdida y matrices de confusión.

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import ConfusionMatrixDisplay

ROOT = Path.cwd().resolve()
if ROOT.name == 'src':
    ROOT = ROOT.parent
RESULT_FILES = [ROOT / 'results' / name for name in ('pytorch.json', 'sklearn.json', 'keras.json')]
missing = [path.name for path in RESULT_FILES if not path.exists()]
if missing:
    raise FileNotFoundError(f'Faltan resultados: {missing}. Ejecuta primero los tres notebooks.')
results = [json.loads(path.read_text(encoding='utf-8')) for path in RESULT_FILES]
```

```python
print(f"{'Framework':<16} {'Accuracy':>10} {'Tiempo (s)':>12}")
print('-' * 40)
for result in results:
    print(f"{result['framework']:<16} {result['accuracy']:>9.2%} {result['training_time_seconds']:>12.2f}")
```

```python
plt.figure(figsize=(8, 5))
for result in results:
    epochs = range(1, len(result['loss']) + 1)
    plt.plot(epochs, result['loss'], marker='o', label=result['framework'])
plt.title('Curvas de pérdida por framework')
plt.xlabel('Época')
plt.ylabel('Cross-entropy')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for result, axis in zip(results, axes):
    ConfusionMatrixDisplay(np.asarray(result['confusion_matrix'])).plot(ax=axis, cmap='Blues', colorbar=False)
    axis.set_title(result['framework'])
plt.tight_layout()
plt.show()
```

## Conclusiones

Compara la tabla y las figuras obtenidas. Las diferencias pequeñas de accuracy son normales por la inicialización, el orden de los lotes y detalles internos de cada optimizador.
